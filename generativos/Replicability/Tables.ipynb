```python
# Results are computed from Table 1 model means and the ShapeNet / mirrored-ShapeNet means already loaded from computed_NSCD/*.json.
# Delta_1 = |mu_gt - mu_NSCD_orig| and Delta_2 = |mu'_gt - mu_NSCD_mirr|.
ground_truth_means = {
    "airplane": {"Orig.": airplane_mean, "Mirr.": mirrored_airplane_mean},
    "car": {"Orig.": car_mean, "Mirr.": mirrored_car_mean},
    "chair": {"Orig.": chair_mean, "Mirr.": mirrored_chair_mean},
}

header_groups = [(CATEGORY_LABELS[c], ["Delta_1", "Delta_2"]) for c in CATEGORIES]
table2_rows = []
for model in MODELS:
    values = []
    for category in CATEGORIES:
        delta_1 = abs(ground_truth_means[category]["Orig."] - table1_data[model][category]["Orig."])
        delta_2 = abs(ground_truth_means[category]["Mirr."] - table1_data[model][category]["Mirr."])
        values.append(fmt4(delta_1))
        values.append(fmt4(delta_2))
    table2_rows.append((model, values))

table2 = render_table("Table 2. Symmetry preservation results.", header_groups, table2_rows)
```

## Table 3: Quantitative comparison of the symmetry-breaking inspection


```python
import re

TABLE3_RESULTS_DIR = REPRO_DIR / "table3_results"
if not TABLE3_RESULTS_DIR.exists():
    raise FileNotFoundError(f"Missing Table 3 results directory: {TABLE3_RESULTS_DIR}")


def average_means_from_notebook(notebook_path, expected_values=6):
    notebook_path = Path(notebook_path)
    if not notebook_path.exists():
        raise FileNotFoundError(f"Missing Table 3 result notebook: {notebook_path}")

    notebook = json.loads(notebook_path.read_text(encoding="utf-8"))
    stream_text = []
    for cell in notebook.get("cells", []):
        for output in cell.get("outputs", []):
            if output.get("output_type") == "stream":
                stream_text.append("".join(output.get("text", [])))

    values = [
        float(match.group(1))
        for match in re.finditer(r"Average mean\s+\d+:\s+([0-9.eE+-]+)", "\n".join(stream_text))
    ]
    if len(values) != expected_values:
        raise ValueError(
            f"Expected {expected_values} saved 'Average mean' outputs in {notebook_path}, found {len(values)}. "
            "Run that notebook or check its saved outputs."
        )
    return values


def table3_values_from_notebook(notebook_path):
    values = average_means_from_notebook(notebook_path)
    return {
        "airplane": {"Orig.": values[0], "Mirr.": values[3]},
        "car": {"Orig.": values[1], "Mirr.": values[4]},
        "chair": {"Orig.": values[2], "Mirr.": values[5]},
    }


# Source notebooks for Table 3: Replicability/table3_results/*.ipynb
table3_sources = {
    "DPM": TABLE3_RESULTS_DIR / "DPM_ComputeCD.ipynb",
    "PVD": TABLE3_RESULTS_DIR / "PVD_ComputeCD.ipynb",
    "DiT-3D": TABLE3_RESULTS_DIR / "DiT-3D_ComputeCD.ipynb",
    "XCube": TABLE3_RESULTS_DIR / "XCube_ComputeCD.ipynb",
    "SLIDE 3D": TABLE3_RESULTS_DIR / "SLIDE3D_ComputeCD.ipynb",
    "SPVD-S": TABLE3_RESULTS_DIR / "SPVD-S_ComputeCD.ipynb",
    "SPVD-L": TABLE3_RESULTS_DIR / "SPVD-L_ComputeCD.ipynb",
    "LION (E)": TABLE3_RESULTS_DIR / "GenerateLatent_fromVAE.ipynb",
}

table3_data = {model: table3_values_from_notebook(path) for model, path in table3_sources.items()}

header_groups = [(CATEGORY_LABELS[c], ["Orig.", "Mirr."]) for c in CATEGORIES]
table3_rows = []
for model in table3_sources:
    values = []
    for category in CATEGORIES:
        values.append(fmt4(table3_data[model][category]["Orig."]))
        values.append(fmt4(table3_data[model][category]["Mirr."]))
    table3_rows.append((model, values))

table3 = render_table("Table 3. Quantitative comparison of the symmetry-breaking inspection.", header_groups, table3_rows)
```

## Table 4: Generation results on Airplane, Car, and Chair compared with original models’ results using 1-NNA↓ and COV↑ as metrics. 


```python
import json
from html import escape
from pathlib import Path
from IPython.display import HTML, display

TABLE4_RESULTS_DIR = REPRO_DIR / "metric_results"
TABLE4_FILES = {
    "Original": TABLE4_RESULTS_DIR / "original_models.json",
    "HM": TABLE4_RESULTS_DIR / "mirrored-object_models.json",
    "S": TABLE4_RESULTS_DIR / "symmetrical_models.json",
}
TABLE4_FAMILY_ORDER = ["DPM", "PVD", "LION", "DiT-3D", "XCube", "SLIDE 3D", "SPVD-S", "SPVD-L"]
TABLE4_FAMILY_TO_KEY = {
    "DPM": "dpm",
    "PVD": "pvd",
    "LION": "lion",
    "DiT-3D": "dit3d",
    "XCube": "xcube",
    "SLIDE 3D": "slide3d",
    "SPVD-S": "spvd_s",
    "SPVD-L": "spvd_l",
}


def load_table4_results(json_path):
    results = json.loads(Path(json_path).read_text(encoding="utf-8"))
    grouped = {}
    for sample_path, metrics in results.items():
        stem = Path(sample_path).stem
        if stem.startswith("generated_ours_"):
            suffix = stem[len("generated_ours_"):]
        elif stem.startswith("generated_"):
            suffix = stem[len("generated_"):]
        elif stem.startswith("ablation_"):
            suffix = stem[len("ablation_"):]
        else:
            raise ValueError(f"Unexpected sample name in Table 4 data: {sample_path}")
        family_key, category = suffix.rsplit("_", 1)
        grouped.setdefault(family_key, {})[category] = metrics
    return grouped


def table4_cell(metrics, key):
    return fmt2(metrics[key] * 100.0)


variant_results = {variant: load_table4_results(path) for variant, path in TABLE4_FILES.items()}

table4_rows = []
for family_label in TABLE4_FAMILY_ORDER:
    family_key = TABLE4_FAMILY_TO_KEY[family_label]
    for variant_label, row_prefix in [("Original", ""), ("HM", "HM-"), ("S", "S-")]:
        family_results = variant_results[variant_label][family_key]
        row_name = f"{row_prefix}{family_label}"
        row_values = []
        for category in CATEGORIES:
            metrics = family_results[category]
            row_values.extend([
                table4_cell(metrics, "1-NN-CD-acc"),
                table4_cell(metrics, "1-NN-EMD-acc"),
                table4_cell(metrics, "lgan_cov-CD"),
                table4_cell(metrics, "lgan_cov-EMD"),
            ])
        table4_rows.append((row_name, row_values, variant_label, family_label))

table4_html = [
    TABLE_STYLE,
    "<style>",
    ".paper-table thead tr:nth-child(1) th { border-bottom: 1px solid #ddd; }",
    ".paper-table thead tr:nth-child(2) th { border-bottom: 1px solid #ddd; }",
    ".paper-table thead tr:nth-child(3) th { border-bottom: 2px solid #222; }",
    ".paper-table tbody tr.group-start td { border-top: 2px solid #444; }",
    ".paper-table tbody tr.variant-s td:first-child { font-style: italic; }",
    "</style>",
    "<table class='paper-table'>",
    "<caption>Table 4. Generation results on Airplane, Car, and Chair compared with original models' results using 1-NNA↓ and COV↑ as metrics.</caption>",
    "<thead>",
    "<tr><th rowspan='3'>Model</th>",
    "".join(f"<th colspan='4'>{CATEGORY_LABELS[category]}</th>" for category in CATEGORIES),
    "</tr>",
    "<tr>",
    "".join("<th colspan='2'>1-NNA (↓)</th><th colspan='2'>COV (↑)</th>" for _ in CATEGORIES),
    "</tr>",
    "<tr>",
    "".join("<th>CD</th><th>EMD</th><th>CD</th><th>EMD</th>" for _ in CATEGORIES),
    "</tr>",
    "</thead><tbody>",
]

for row_name, row_values, variant_label, family_label in table4_rows:
    row_classes = [f"variant-{variant_label.lower()}"]
    if variant_label == "Original" and family_label != TABLE4_FAMILY_ORDER[0]:
        row_classes.append("group-start")
    class_attr = " ".join(row_classes)
    table4_html.append(f"<tr class='{class_attr}'><td>{escape(row_name)}</td>")
    for value in row_values:
        table4_html.append(f"<td>{value}</td>")
    table4_html.append("</tr>")

table4_html.append("</tbody></table>")
display(HTML("".join(table4_html)))

table4 = {
    "title": "Table 4. Generation results on Airplane, Car, and Chair compared with original models' results using 1-NNA↓ and COV↑ as metrics.",
    "rows": table4_rows,
}
```
