```python
import sys
import statistics
from pathlib import Path

import numpy as np
import torch

PROJECT_ROOT = Path.cwd()
if PROJECT_ROOT.name == "table3_results":
    PROJECT_ROOT = PROJECT_ROOT.parent.parent
elif PROJECT_ROOT.name == "Replicability":
    PROJECT_ROOT = PROJECT_ROOT.parent

REPLICABILITY_DIR = PROJECT_ROOT / "Replicability"
SAMPLES_TABLE3_DIR = REPLICABILITY_DIR / "samples_table3"
SYMMETRY_PROTOCOL_DIR = PROJECT_ROOT / "Symmetry_Measurement_Protocol"
if str(SYMMETRY_PROTOCOL_DIR) not in sys.path:
    sys.path.append(str(SYMMETRY_PROTOCOL_DIR))

from ChamferDistance import chamfer_distance
```

**Replicability Note:**

Some values may differ slightly when this notebook is rerun, as the point-cloud preparation process uses a point-sampling algorithm that may introduce minor variations.

# Original dataset

## Airplane

```python
one_airplane_input_original_objects = SAMPLES_TABLE3_DIR / "SPVD-L" / "original_SPVD-L_airplane_original.pth"
one_airplane_input_mirrored_objects = SAMPLES_TABLE3_DIR / "SPVD-L" / "original_SPVD-L_airplane_mirrored.pth"

one_airplane_files_original_objects = torch.load(one_airplane_input_original_objects).cpu().numpy()
one_airplane_files_mirrored_objects = torch.load(one_airplane_input_mirrored_objects).cpu().numpy()

one_cd_results_airplane = []
for i in range(len(one_airplane_files_original_objects)):
    original_pc = one_airplane_files_original_objects[i]
    mirrored_pc = one_airplane_files_mirrored_objects[i]

    if original_pc.shape[0] == 15000:
        pass

    cd_result = chamfer_distance(original_pc, mirrored_pc)
    one_cd_results_airplane.append(cd_result)

print("\nAverage mean " + str(len(one_airplane_files_original_objects)) + ":", statistics.mean(one_cd_results_airplane))
print("Average mean/max:", max(one_cd_results_airplane))
print("Average mean/min:", min(one_cd_results_airplane))
```
